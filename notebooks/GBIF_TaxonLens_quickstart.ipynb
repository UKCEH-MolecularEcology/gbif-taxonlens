## Tips

- For DADA2-style files, columns such as `Kingdom`, `Phylum`, `Class`, `Order`, `Family`, `Genus`, and `Species` are detected automatically.
- Prefixes such as `k__`, `p__`, `g__`, and `s__` are cleaned before matching.
- If `Species` contains only an epithet, for example `s__robur`, TaxonLens combines it with `Genus`, for example `Quercus robur`.
- Rows with `FUZZY`, `HIGHERRANK`, `AGGREGATE`, or `NONE` should be reviewed manually.
